
## Krótkie wprowadzenie do liczb zespolonych

W szkole zwykle spotykamy tylko liczby rzeczywiste – takie, które można zaznaczyć na zwykłej osi liczbowej. Liczby zespolone to naturalne rozszerzenie tego pomysłu. Zamiast jednej osi używamy dwóch osi - rzeczywistej i urojonej.
Liczby zespolone rozszerzają liczby rzeczywiste o element $i$, który spełnia:
$ i^2 = -1 $.

Każdą liczbę zespoloną zapisujemy jako
$ z = x + iy $,  
gdzie $x$ i $y$ są liczbami rzeczywistymi.  
- $x$ nazywamy **częścią rzeczywistą**,  
- $y$ nazywamy **częścią urojoną**.

---

## Działania na liczbach zespolonych

Dla dwóch liczb  
$ z_1 = x_1 + i y_1 $,  
$ z_2 = x_2 + i y_2 $,  
definiujemy:

### Dodawanie
$ z_1 + z_2 = (x_1 + x_2) + i(y_1 + y_2) $.

### Odejmowanie
$ z_1 - z_2 = (x_1 - x_2) + i(y_1 - y_2) $.

### Mnożenie
$ z_1 \cdot z_2 = (x_1 x_2 - y_1 y_2) + i(x_1 y_2 + x_2 y_1) $.

---

## Moduł i argument liczby zespolonej

### Moduł
Długość wektora reprezentującego liczbę $z = x + iy$ na płaszczyźnie:
$ |z| = \sqrt{x^2 + y^2} $.

### Argument
Argumentem liczby zespolonej $z = x + iy$ nazywamy kąt, jaki wektor $(x, y)$ tworzy z dodatnią osią $x$:
$ \arg z = \theta $.


## Liczby zespolone w Pythonie (NumPy)

W Pythonie liczby zespolone zapisujemy w postaci  
`x + y*1j`,  
gdzie `1j` pełni rolę jednostki urojonej $i$.

Przykład:  
- $3 + 4i$ zapisujemy jako `3 + 4j`.

NumPy potrafi wykonywać działania na liczbach zespolonych automatycznie.

### Przykład użycia


````python
import numpy as np

# Definicja liczb zespolonych
z1 = 3 + 4j
z2 = -1 + 2j

# Podstawowe działania
suma = z1 + z2
roznica = z1 - z2
iloczyn = z1 * z2

# Moduł i argument
modul = np.abs(z1)
arg = np.angle(z1)

print("z1 =", z1)
print("z2 =", z2)
print("Suma =", suma)
print("Różnica =", roznica)
print("Iloczyn =", iloczyn)
print("Moduł z1 =", modul)
print("Argument z1 =", arg)
````

## Szeregi potęgowe w liczbach zespolonych

Szeregi potęgowe działają w liczbach zespolonych dokładnie tak samo jak w liczbach rzeczywistych.  
Podstawowa idea jest taka sama: funkcję można przedstawić jako **sumę nieskończonego szeregu wyrazów** zależnych od liczby zespolonej $z$.

### Definicja

Szereg potęgowy wokół $z=0$ ma postać:

$$
f(z) = \sum_{n=0}^{\infty} c_n z^n = c_0 + c_1 z + c_2 z^2 + c_3 z^3 + \dots
$$

- $c_n$ – współczynniki (mogą być rzeczywiste lub zespolone)  
- $z^n$ – potęga liczby zespolonej

Częściowa suma pierwszych $N+1$ wyrazów:

$$
S_N(z) = \sum_{n=0}^{N} c_n z^n
$$

jest **przybliżeniem funkcji** $f(z)$.  
Im większe $N$, tym dokładniejsze przybliżenie.

---

### Przykład: funkcja wykładnicza

Funkcję $e^z$ można zapisać jako szereg potęgowy:

$$
e^z = 1 + z + \frac{z^2}{2!} + \frac{z^3}{3!} + \frac{z^4}{4!} + \dots
$$

- Częściowa suma $S_0(z) = 1$ – bardzo słabe przybliżenie  
- Częściowa suma $S_3(z) = 1 + z + \frac{z^2}{2} + \frac{z^3}{6}$ – już lepiej  
- Większe $N$ → dokładniejsze przybliżenie

---

### Co warto zauważyć

1. Szeregi potęgowe można liczyć dla **dowolnej liczby zespolonej** $z = x + iy$.  
2. Każdy wyraz szeregu jest również liczbą zespoloną.  
3. Częściowe sumy są łatwe do policzenia w Pythonie i NumPy.

## Częściowe sumy szeregów potęgowych

W tym ćwiczeniu Twoim zadaniem jest **samodzielnie napisać funkcje**, które liczą częściowe sumy szeregu potęgowego dla liczb zespolonych.

### Co musisz zrobić

1. Napisz trzy funkcje:

- `exp_series(z, N)` – oblicza częściową sumę szeregu potęgowego funkcji $e^z$ dla liczby zespolonej `z` i `N` wyrazów.
- `sin_series(z, N)` – oblicza częściową sumę szeregu potęgowego funkcji $\sin z$.
- `cos_series(z, N)` – oblicza częściową sumę szeregu potęgowego funkcji $\cos z$.

2. Każda funkcja powinna działać zarówno dla **pojedynczej liczby zespolonej**, jak i dla **tablicy liczb zespolonych** (NumPy).

3. W częściowej sumie należy **iteracyjnie dodawać wyrazy szeregu**:
   - dla $e^z$: $1 + z + z^2/2! + z^3/3! + \dots$
   - dla $\sin z$: $z - z^3/3! + z^5/5! - z^7/7! + \dots$
   - dla $\cos z$: $1 - z^2/2! + z^4/4! - z^6/6! + \dots$

### Wskazówki

- Możesz użyć zmiennej `term`, żeby przechowywać kolejny wyraz szeregu.  
- Pamiętaj o znaku przy funkcjach sinus i cosinus.  

---

**Komentarz dla leniwych:** rozwiązanie, które się nasuwa, to napisanie 3 podobnych funkcji. Jeśli jednak nie chce Ci się tego robić, poczytaj o wzorze Eulera.

````python
def exp_series(z, N):
    """
    Częściowa suma szeregu potęgowego funkcji e^z
    z : liczba zespolona lub tablica liczb zespolonych
    N : liczba wyrazów szeregu
    """
    S = 0
    term = 1  # pierwszy wyraz = z^0 / 0! = 1
    for n in range(N+1):
        S += term
        term *= z / (n+1)
    return S



def cos_series(z, N):

    return 0.5*(exp_series(1j*z,N)+exp_series((-1)*1j*z,N))


def sin_series(z, N):

    return -1j*0.5*(exp_series(1j*z,N)-exp_series((-1)*1j*z,N))

````

## Kolorowa wizualizacja funkcji zespolonych: system HSV

W tej wizualizacji będziemy korzystać z modelu barw **HSV**, który składa się z trzech parametrów:

1. **H – Hue (barwa)**  
   - Określa "rodzaj koloru", np. czerwony, zielony, niebieski.  
   - W naszym przypadku będzie odpowiadać **argumentowi liczby zespolonej** (kąt w płaszczyźnie zespolonej).  
   - Zakres: od 0 do 1 (całe koło kolorów).

   128px-Color_circle_(RGB).svg.png

2. **S – Saturation (nasycenie)**  
   - Określa intensywność koloru: 0 → szaro, 1 → pełny kolor.  
   - W wizualizacji ustawiamy stałe `S = 1`, żeby kolory były wyraźne.

3. **V – Value / Brightness (jasność)**  
   - Określa jasność koloru: 0 → czarny, 1 → pełna jasność.  
   - W naszym przypadku będzie odpowiadać **modułowi liczby zespolonej** (długości wektora) i skalujemy go logarytmicznie, żeby duże wartości nie dominowały obrazu.

---

### Dlaczego HSV jest wygodne w wizualizacji liczb zespolonych

- Argument liczby zespolonej jest naturalnie kątem → pasuje do barwy (H).  
- Moduł liczby zespolonej jest naturalnie długością wektora → pasuje do jasności (V).  
- W ten sposób w jednej mapie kolorów widzimy zarówno **kierunek**, jak i **długość** liczby zespolonej.
---

## Instrukcja krok po kroku: jak działa `plot_complex_map`

Funkcja `plot_complex_map(W, title)` tworzy efektowną wizualizację funkcji zespolonej. Oto, co dokładnie się dzieje:

1. **Argument i moduł liczby zespolonej**  
   -  obliczamy kąt liczby zespolonej w radianach (od -π do π).  
   -  obliczamy długość wektora zespolonego (moduł).

2. **Normalizacja argumentu do barwy (Hue, H)**  
   - Tutaj musimy zamienić zakres $[-\pi,\pi]$ na zakres $[0,1]$ tak, żeby odpowiadał pełnemu kołu kolorów. Przypominamy, że umiemy już określić bijekcję $f:[a,b] \to [c,d]$ jako
   $$f(x)=(d-c)\frac{x-a}{b-a}+c.$$
   Wykorzystaj tę bijekcję w kodzie.

3. **Normalizacja modułu do jasności (Value, V)**  
   - `V = np.log1p(modulus)` – logarytmicznie skalujemy moduł, aby duże wartości nie dominowały obrazu.  
   - `V = V / np.max(V)` – przeskalowanie do zakresu [0, 1].

4. **Nasycenie koloru (Saturation, S)**  
   - `S = np.ones_like(H)` – ustawiamy nasycenie na 1, aby kolory były intensywne.

5. **Połączenie HSV i konwersja do RGB**  
   - `HSV = np.stack((H, S, V), axis=-1)` – tworzy tablicę z trzema kanałami HSV.  
   - `RGB = hsv_to_rgb(HSV)` – konwertuje kolory HSV na RGB, które można wyświetlić w matplotlib.

6. **Rysowanie mapy**  
   - `plt.imshow(RGB, origin='lower', extent=[x_min, x_max, y_min, y_max])` – wyświetla obrazek z poprawną orientacją osi.  
   - `plt.xlabel`, `plt.ylabel`, `plt.title` – podpisy osi i tytuł.  
   - `plt.show()` – wyświetlenie mapy.

---

### Efekt wizualny

- **Barwa (Hue, H)** pokazuje **argument liczby zespolonej** (kąt).  
- **Jasność (Value, V)** pokazuje **moduł liczby zespolonej** (długość wektora).  
- Możesz zmieniać funkcję (`W_exp`, `W_sin`, `W_cos`) i liczbę wyrazów `N`, aby obserwować różne wzory na płaszczyźnie zespolonej.

````python
def plot_complex_map(W, title="Funkcja zespolona"):
    """
    Tworzy kolorową mapę funkcji zespolonej W
    - barwa (hue) = argument liczby zespolonej
    - jasność (value) = moduł liczby zespolonej (logarytmicznie skalowany)
    """
    # Oblicz argument i moduł
    arg = np.angle(W)        # od -pi do pi
    modulus = np.abs(W)

    # Normalizacja argumentu do zakresu [0, 1] dla barwy
    H = (arg + np.pi) / (2 * np.pi)

    # Normalizacja modułu do zakresu [0, 1] dla jasności (logarytmicznie)
    V = np.log1p(modulus)
    V = V / np.max(V)

    # Nasycenie ustalone na 1
    S = np.ones_like(H)

    # Połączenie HSV i konwersja na RGB
    HSV = np.stack((H, S, V), axis=-1)
    RGB = hsv_to_rgb(HSV)

    # Rysowanie mapy
    plt.figure(figsize=(10,10))
    plt.imshow(RGB, origin='lower', extent=[x_min, x_max, y_min, y_max])
    plt.xlabel("Re(z)")
    plt.ylabel("Im(z)")
    plt.title(title)
    plt.show()


# Przykład dla funkcji wykładniczej
plot_complex_map(W_exp)
````
