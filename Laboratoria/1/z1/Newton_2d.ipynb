Jeśli wektor gradientu $\nabla f$ mówi nam, w którą stronę iść, to macierz pochodnych drugiego rzędu – **Macierz Hessa (Hesian)** – mówi nam, *jak bardzo ten gradient za chwilę się zmieni*. Hesian mierzy lokalną krzywiznę naszej funkcji. Więcej teorii będzie podane na najbliższych wykładach.

$$H=\begin{bmatrix}\frac{\partial^2 f}{\partial x^2}&\frac{\partial^2 f}{\partial x \partial y}\\ \frac{\partial^2 f}{\partial y \partial x} & \frac{\partial^2 f}{\partial y^2}\end{bmatrix}$$

Zamiast przykładać do zbocza płaską kartkę (płaszczyznę styczną), algorytm użyje Hesianu do zbudowania wirtualnej wielowymiarowej "miski" (paraboloidy), która idealnie dopasuje się do lokalnych krzywizn naszej góry (to tzw. aproksymacja Taylora drugiego rzędu).

Ten genialny skok, polegający na przemnożeniu gradientu przez **odwrotność macierzy Hessa** ($H^{-1}$), to fundament **Metody Newtona** w optymalizacji.

### Metoda Newtona: Od ślepego błądzenia do zaginania przestrzeni

Aby w pełni docenić potęgę Metody Newtona, musimy na chwilę wrócić do Spadku Gradientowego. Pamiętasz, jak działał? Znał tylko lokalne nachylenie terenu (pochodną pierwszego rzędu). Z matematycznego punktu widzenia, Spadek Gradientowy buduje w danym punkcie **płaszczyznę styczną** do funkcji i zsuwa się po niej o mały krok ($\alpha$).

Problem polega na tym, że prawdziwy świat rzadko jest płaski. W "wąskich dolinach" płaszczyzna styczna drastycznie mija się z prawdziwym kształtem funkcji już ułamek milimetra dalej. Algorytm wpada w ścianę i zaczyna oscylować.

Tutaj na scenę wkracza **Metoda Newtona**.

Zamiast udawać, że świat jest płaski, Metoda Newtona patrzy na **Macierz Hessa (Hesian)**. Ponieważ Hesian zawiera informacje o pochodnych drugiego rzędu (czyli o krzywiźnie, wypukłościach i wklęsłościach), algorytm potrafi zbudować lokalnie coś znacznie lepszego niż płaska kartka. Buduje **wielowymiarową paraboloidę** (wirtualną miskę), która "przytula się" do naszej funkcji, idealnie naśladując jej lokalny kształt.

#### Matematyczne Serce: Szereg Taylora

Wszystko opiera się na przybliżeniu naszej funkcji $f(x)$ w okolicy punktu $x_n$ za pomocą Szeregu Taylora drugiego rzędu. Dla wektorów wygląda to tak:

$$f(x)\approx f(x_n)+\nabla f(x_n)^T(x-x_n)+\frac{1}{2}(x-x_n)^TH(x-x_n)$$

Gdzie:
* $\nabla f(x_n)$ to wektor gradientu (nachylenie).
* $H$ to macierz Hessa (krzywizna).
* $x-x_n$ to nasz krok (wektor przesunięcia).

Zauważcie, co tu się stało: zastąpiliśmy naszą skomplikowaną, nieliniową funkcję straty piękną, kwadratową paraboloidą.

#### Magia Teleportacji (Wyprowadzenie wzoru)

Skoro masz teraz funkcję kwadratową (wirtualną miskę), to nie musimy po niej powoli schodzić! Z algebry i analizy wiesz, że funkcja kwadratowa ma swoje ekstremum tam, gdzie jej pochodna wynosi zero.

Zróżnicujmy więc nasze przybliżenie Taylora po $x$ i przyrównajmy do zera, aby znaleźć absolutne dno tej miski:

$$\nabla f(x_n)+H(x-x_n)=0$$

Teraz wystarczy przekształcić to równanie, aby wyznaczyć nasz nowy punkt $x$ (który nazwiemy $x_{n+1}$):

$$H(x_{n+1}-x_n)=-\nabla f(x_n)$$

Mnożymy lewostronnie przez odwrotność macierzy Hessa ($H^{-1}$) i przenosimy $x_n$:

$$x_{n+1}=x_n-H^{-1}\nabla f(x_n)$$

I to jest właśnie **wzór aktualizacji Metody Newtona**.

Zauważ coś niesamowitego: **nie ma tu współczynnika uczenia ($\alpha$)!** Algorytm sam wie, jak duży krok postawić. Po prostu wylicza, gdzie znajduje się środek lokalnej krzywizny i teleportuje się tam w jednym kroku. Dla funkcji czysto kwadratowych (jak nasza testowa dolina $f(x,y)=x^2+20y^2$) Metoda Newtona dociera do absolutnego minimum zawsze w **dokładnie jednej iteracji**, niezależnie od tego, jak wąska jest ta dolina i skąd wystartujemy.

#### Haczyk: Dlaczego nie używasz tego wszędzie w Deep Learningu?

Skoro to taka rewelacja, dlaczego do trenowania głębokich sieci neuronowych używamy takich algorytmów jak Adam czy SGD, a nie Metody Newtona? Są dwa kluczowe powody, o których każdy inżynier AI musi pamiętać:

1. **Koszt obliczeniowy i pamięciowy (Klątwa Wymiarowości):** W modelu językowym mającym miliard parametrów, gradient to wektor o rozmiarze miliarda elementów. Da się to policzyć. Ale Hesian? To macierz o rozmiarze miliard na miliard! Odwrócenie takiej macierzy w każdym kroku (złożoność rzędu $O(N^3)$) stopiłoby największe superkomputery na świecie.
2. **Przyciąganie przez punkty siodłowe:** Metoda Newtona szuka miejsca, gdzie gradient wynosi zero. Nie obchodzi jej, czy to minimum, maksimum, czy punkt siodłowy. Jeśli znajdzie się w okolicy punktu siodłowego (gdzie Hesian nie jest dodatnio określony), potrafi radośnie polecieć w stronę siodła lub wręcz pod górę, zamiast spadać w dół.

### Krok 5: Wielka Ucieczka – Saddle-Free Newton i Wartości Własne

Widziałeś/aś to na własne oczy w poprzednim eksperymencie z funkcją $f(x,y) = x^2 - y^2$. Zamiast zjechać w dół (w kierunku ujemnych wartości na osi Y), nasz wspaniały algorytm z radością wskoczył na sam środek przełęczy, w punkt (0,0), i uznał, że jego praca jest skończona.

**Dlaczego to zrobił?**
Metoda Newtona opiera się na prostym założeniu: szuka miejsca, gdzie gradient jest równy zero ($\nabla f = 0$). Problem polega na tym, że gradient zeruje się w trzech przypadkach:
1. W lokalnym minimum (dno doliny).
2. W lokalnym maksimum (szczyt góry).
3. W **punkcie siodłowym** (środek przełęczy).

Ponieważ algorytm buduje swoją "wirtualną miskę" używając Macierzy Hessa, w punkcie siodłowym ta miska wcale nie jest miską! Zyskuje kształt czipsa Pringles – w jednym kierunku zakrzywia się do góry, a w innym w dół.



Aby Metoda Newtona zawsze szła w stronę minimum, potrzebujemy gwarancji, że nasz "wirtualny model" terenu jest zawsze uśmiechniętą, wypukłą miską. Z punktu widzenia algebry liniowej oznacza to, że Macierz Hessa musi być **dodatnio określona** (Positive Definite).

**Rozwiązanie: Inżynieria Wartości Własnych (Eigenvalue Engineering)**

Skoro znasz zaawansowaną algebrę liniowej, to wiesz, że każdą symetryczną macierz (a Hesian jest symetryczny z mocy Twierdzenia Schwarza!) możesz rozłożyć na wektory własne i wartości własne:

$$H = V \Lambda V^T$$

Gdzie:
* $V$ to macierz ortogonalna zawierająca wektory własne (wskazujące główne kierunki krzywizny – np. wzdłuż i w poprzek siodła).
* $\Lambda$ to macierz diagonalna zawierająca wartości własne (mówiące o tym, jak bardzo i w którą stronę teren się zakrzywia).

**Wartości własne to nasz klucz do sukcesu:**
* Jeśli wartość własna jest dodatnia $\lambda > 0$, teren w tym kierunku przypomina miskę (chcemy tam iść).
* Jeśli wartość własna jest ujemna $\lambda < 0$, teren przypomina górę (zamiast spadać, Newton będzie chciał się na nią wspiąć!).

**Algorytm Saddle-Free Newton (Newton Odporny na Siodła):**
Zamiast akceptować Hesian takim, jakim jest, "zhakujemy" go. Wyciągniemy z niego wartości własne, wymusimy, aby wszystkie były dodatnie (np. wyciągając z nich wartość bezwzględną), i poskładamy macierz z powrotem!

Nowy, zmodyfikowany Hesian będzie zdefiniowany tak:

$$H_{mod} = V |\Lambda| V^T$$

Dzięki temu algorytm zachowa informację o kierunkach (wektory własne się nie zmieniają), ale potraktuje kierunek spadkowy jak "ścianę miski", od której musi się odbić. Zamiast utknąć na siodle, z pełną prędkością zjedzie w dół!

**Co zrobimy w następnej komórce:**
1. Użyjemy funkcji `np.linalg.eigh(H)`, która jest zoptymalizowana do rozkładu macierzy symetrycznych.
2. Zastosujesz wartość bezwzględną na wartościach własnych.
3. Odtworzysz zmodyfikowany Hesian i zaktualizujemy pozycję.

Zanim napiszesz pętlę dla Saddle-Free Newtona, musisz zrozumieć Twoje główne narzędzie: funkcję z biblioteki NumPy do rozkładu widmowego macierzy.

W dokumentacji znajdziesz dwie podobne funkcje: `np.linalg.eig` oraz `np.linalg.eigh`. Różnica polega na jednej literce. Litera **"h"** oznacza **Hermitian** (macierz hermitowska). W świecie liczb rzeczywistych oznacza to po prostu **macierz symetryczną**. Ponieważ nasz Hesian (macierz drugich pochodnych) jest zawsze symetryczny, używamy szybszej i stabilniejszej wersji `eigh`. Daje nam to matematyczną gwarancję, że wyniki będą liczbami rzeczywistymi, a wyznaczone kierunki będą do siebie prostopadłe.



Kiedy wywołasz funkcję `w, v = np.linalg.eigh(H)`, algorytm weźmie Twój Hesian i rozbije go na dwa elementy:

**1. Tablica `w` (Wartości własne / Eigenvalues):**
* Jest to jednowymiarowa lista liczb (skalarów) oznaczanych jako $\lambda$.
* **Co to oznacza:** Te liczby mówią nam, jak bardzo teren zakrzywia się w głównych kierunkach. Liczba dodatnia to zakrzywienie w górę (miska), ujemna to zakrzywienie w dół (zbocze).
* **Uwaga:** NumPy domyślnie sortuje te wartości od najmniejszej do największej.

**2. Macierz `v` (Wektory własne / Eigenvectors):**
* Jest to dwuwymiarowa macierz oznaczana jako $V$.
* **Co to oznacza:** Te wektory wyznaczają nowe, idealnie dopasowane osie naszego terenu (np. oś biegnącą dokładnie wzdłuż dna doliny).
*  **Największa pułapka programistyczna:** Wektory własne są ułożone w **KOLUMNACH**, a nie w wierszach! Kolumna o indeksie `i` odpowiada wartości własnej `w[i]` .
