Minimalizując $L(x)$, znajdziemy dawkę, która najlepiej pasuje do wszystkich osób.

---

### Funkcja straty

Funkcja straty definiowana jest jako średni kwadrat różnicy między rzeczywistym stężeniem $y_i$ a przewidywanym stężeniem przy dawce $x$:

$$
L(x) = \frac{1}{100} \sum_{i=1}^{100} \Big( y_i - 10 \cdot e^{-0.01(x - x_i)^2} \Big)^2
$$

- $x_i$ – dawka, którą faktycznie podano i-tej osobie  
- $y_i$ – zmierzone stężenie witaminy we krwi i-tej osoby (mg/dL)  
- $x$ – dawka, którą chcemy dobrać  

Minimalizując $L(x)$, znajdziemy dawkę witaminy X, która maksymalizuje średnie stężenie we krwi wszystkich uczestników.

$$

$$
