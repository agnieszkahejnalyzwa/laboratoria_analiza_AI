# Zadanie 1: Dystans to pojęcie względne (normy i kule)
# Wprowadzenie teoretyczne

W uczeniu maszynowym dane reprezentujemy jako punkty (wektory) w przestrzeni $\mathbb{R}^n$. Aby stwierdzić, czy dwa obiekty są do siebie podobne (np. czy dany obrazek to ta sama cyfra, lub czy klient banku zachowuje się podejrzanie), musimy umieć mierzyć odległość między nimi.

Odległość definiujemy za pomocą normy. Najpopularniejsze w AI to:

**Norma euklidesowa ($L_2$)** Klasyczna odległość "w linii prostej".
$$ \|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2} $$

**Norma taksówkowa ($L_1$, Manhattan)** Suma modułów współrzędnych. Kluczowa w metodach takich jak Lasso (dobór cech).
$$ \|x\|_1 = \sum_{i=1}^n |x_i| $$

**Norma maksimum ($L_\infty$, Czebyszewa)** Największa różnica na pojedynczej współrzędnej.
$$ \|x\|_\infty = \max_i |x_i| $$

# Cel zadania

Zaimplementujesz powyższe normy i zobaczysz, jak definicja "jednostkowej odległości od zera" drastycznie zmienia kształt kuli otwartej $B(0,1)$ w zależności od wybranej metryki.


## Krok 1: Implementacja norm

W poniższej komórce uzupełnij kod funkcji obliczających normy dla wektora $x$. Skorzystaj z biblioteki numpy.



```python
import numpy as np
import matplotlib.pyplot as plt

def norma_l1(x):
    """Suma wartości bezwzględnych"""
    # UZUPEŁNIJ: użyj np.abs i np.sum
    return np.sum(np.abs(x))

def norma_l2(x):
    """Pierwiastek sumy kwadratów"""
    # UZUPEŁNIJ: użyj np.sqrt i np.sum kwadratów
    return np.sqrt(np.sum(x**2))

def norma_linf(x):
    """Maksimum z wartości bezwzględnych"""
    # UZUPEŁNIJ: użyj np.max i np.abs
    return np.max(np.abs(x))

# TEST: Dla wektora [3, -4] normy powinny wynosić: L1=7, L2=5, Linf=4
test_v = np.array([3, -4])
print(f"Test L1: {norma_l1(test_v)} (oczekiwano 7)")
print(f"Test L2: {norma_l2(test_v)} (oczekiwano 5)")
print(f"Test Linf: {norma_linf(test_v)} (oczekiwano 4)")
```

## Krok 2: Wizualizacja "Świata według normy"

Teraz sprawdzimy, jak wyglądają kule jednostkowe $B(0,1)=\{x \in \mathbb{R}^2\::\: \|x\| \leq 1 \}$. W geometrii euklidesowej spodziewamy się koła, ale czy w innych normach też?

Uruchom poniższy kod, aby wygenerować wykresy. Zwróć uwagę na to, jak zmienia się obszar, który algorytm uznaje za "bliski zeru".

```python
# Generujemy siatkę punktów na płaszczyźnie
x = np.linspace(-1.5, 1.5, 400)
y = np.linspace(-1.5, 1.5, 400)
X, Y = np.meshgrid(x, y)

# Przygotowanie wykresów
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
normy = [norma_l1, norma_l2, norma_linf]
tytuly = ['Kula w normie L1 (Manhattan)', 'Kula w normie L2 (Euklides)', 'Kula w normie L_inf (Maximum)']

for i, norma in enumerate(normy):
    # Obliczamy normę dla każdego punktu siatki
    # Z-shape to macierz wartości normy dla współrzędnych X i Y
    Z = np.array([norma(np.array([xx, yy])) for xx, yy in zip(np.ravel(X), np.ravel(Y))])
    Z = Z.reshape(X.shape)

    # Rysujemy kontur, gdzie norma == 1 (brzeg kuli) oraz wypełnienie (wnętrze kuli)
    axs[i].contourf(X, Y, Z, levels=[0, 1], colors=['#a1c9f4'], alpha=0.8)
    axs[i].contour(X, Y, Z, levels=[1], colors=['blue'], linewidths=2)
    axs[i].set_title(tytuly[i])
    axs[i].set_aspect('equal')
    axs[i].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

Analiza wyników (Do przemyślenia)

1. Dlaczego "koło" w normie $L^1$ jest rombem? W tej metryce poruszamy się tylko "pionowo i poziomo" (jak taksówka w Nowym Jorku). Aby dojść do punktu (1,0) lub (0.5, 0.5), zużywamy taką samą "odległość" równą 1.

2. Zastosowanie w AI:  W normie $L^1$ "rogi" kuli dotykają osi układu współrzędnych. To sprawia, że algorytmy optymalizacyjne (jak Lasso) chętniej zerują zbędne cechy (parametry), co pozwala na automatyczny wybór najważniejszych danych.

      -W normie $L^2$​ kara za duże błędy rośnie kwadratowo, co sprawia, że algorytm bardzo nie lubi pojedynczych, dużych pomyłek (tzw. outliers).

# Zadanie 2: Gdzie kończy się obiekt? (Brzeg zbioru w prawdziwym świecie)
##Wprowadzenie teoretyczne

Poznaliśmy pojęcie brzegu zbioru ($\partial A$). Zgodnie z formalną definicją topologiczną, punkt $x$ należy do brzegu zbioru $A$, jeśli w każdej (nawet najmniejszej) kuli otwartej $B(x,r)$ znajdziemy:

  -Przynajmniej jeden punkt należący do zbioru $A$.

  -Przynajmniej jeden punkt nienależący do zbioru $A$ (czyli należący do dopełnienia $X \setminus A$).

###Jak to się ma do sztucznej inteligencji i Computer Vision?
Wyobraź sobie zdjęcie z kamery – np. drona, autonomicznego samochodu czy mikroskopu medycznego. Chcemy, aby AI znalazło kontury obiektów (np. pieszego na drodze, komórki nowotworowej).
Zbiór $A$ to wszystkie piksele należące do obiektu. Kula $B(x,r)$ o promieniu 1 piksela to po prostu bezpośredni sąsiedzi danego piksela (otoczenie 3x3).

Zgodnie z matematyczną definicją brzegu, piksel leży na krawędzi obiektu, jeśli w jego otoczeniu 3x3 znajduje się zarówno piksel należący do obiektu, jak i piksel będący tłem. Przekształćmy tę czystą matematykę na działający algorytm widzenia maszynowego!

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage.transform import resize

# 1. Pobieramy prawdziwe zdjęcie testowe z wbudowanej biblioteki (Cameraman)
zdjecie_oryginalne = data.camera()

# 2. Zmniejszamy zdjęcie do 150x150.
# Dlaczego? Czysty Python z podwójną pętlą 'for' jest wolny.
# Zmniejszenie obrazu sprawi, że nasz algorytm zadziała w ułamku sekundy.
zdjecie_male = resize(zdjecie_oryginalne, (150, 150), anti_aliasing=True)

# 3. BINARYZACJA: Tworzymy nasz Zbiór A
# Definiujemy, że nasz obiekt to wszystko, co jest bardzo ciemne (wartość < 0.3)
# Tworzymy macierz zer i jedynek: 1 to nasz zbiór A (obiekt), 0 to dopełnienie (tło)
obraz = (zdjecie_male < 0.3).astype(int)

# Podgląd przygotowanych danych
fig, axs = plt.subplots(1, 2, figsize=(12, 5))
axs[0].imshow(zdjecie_male, cmap='gray')
axs[0].set_title("Oryginalne zdjęcie (skala szarości)")
axs[0].axis('off')

axs[1].imshow(obraz, cmap='Greys')
axs[1].set_title("Nasz zbiór A (Obiekt po binaryzacji)")
axs[1].axis('off')

plt.tight_layout()
plt.show()
```

##Krok 2: Detekcja brzegu $\partial A$ (Twoje zadanie - samodzielna implementacja!)


Czas przekuć czystą teorię z rozdziału na działający algorytm detekcji krawędzi (ang. Edge Detection). Tym razem napiszesz cały kod samodzielnie!

Jak mapujemy topologię na macierze?

-Przestrzeń i zbiór: Nasz zbinaryzowany obraz to macierz (zmienna z poprzedniego kroku), w której jedynki (1) należą do naszego zbioru obiektów A, a zera (0) do jego dopełnienia $X \setminus A$ (tła).

-Kula otoczenia B(x,r): Zamiast okrągłej kuli euklidesowej, w dyskretnym świecie pikseli użyjemy kwadratowego otoczenia. Dla każdego piksela o współrzędnych (i, j) wytniemy podmacierz o rozmiarze 3x3 (piksel centralny + jego 8 bezpośrednich sąsiadów).

###Instrukcja do zaprogramowania:
Napisz algorytm, który przeanalizuje obraz i wygeneruje nową macierz z konturami obiektów. Zastosuj poniższe kroki:

1. Pobierz wymiary zbinaryzowanej macierzy obraz.

2. Stwórz nową, pustą macierz wypełnioną zerami o tych samych wymiarach (np. nazwij ją `brzeg_obiektu`). Posłuży do zapisania wyników.

3. Użyj podwójnej pętli for do iteracji po współrzędnych `(i, j)` obrazu.

      -Ważne: Aby kula (sąsiedztwo 3x3) nie wyszła poza ramy obrazu i nie wywołała błędu IndexError, rozpocznij pętle od indeksu 1 i zakończ na rozmiar - 1. Ignorujemy w ten sposób skrajną ramkę obrazu o szerokości 1 piksela.

4. Wewnątrz pętli wytnij otoczenie 3x3 dla bieżącego piksela. Podpowiedź do wycinania macierzy w NumPy: `obraz[i-1 : i+2, j-1 : j+2]`.

5. Zastosuj definicję brzegu: Piksel należy do brzegu $\partial A$, jeśli w jego otoczeniu znajdziemy zarówno elementy zbioru $A$ (jedynki), jak i elementy tła (zera). Jeśli ten warunek jest spełniony (np. minimum w otoczeniu to zero, a maksimum to jeden), wpisz wartość 1 do swojej wynikowej macierzy `brzeg_obiektu` pod adresem `[i, j]`.

```python
rozmiar_x, rozmiar_y = obraz.shape

# Tworzymy pustą, czarną planszę na nasz wynik (kontury)
brzeg_obiektu = np.zeros_like(obraz)

# Iterujemy po obrazie (pomijamy skrajne 1-pikselowe ramki, żeby nie wyjść poza tablicę)
for i in range(1, rozmiar_x - 1):
    for j in range(1, rozmiar_y - 1):

        # Kula B(x, r) - sąsiedztwo 3x3 wokół piksela (i, j)
        otoczenie = obraz[i-1 : i+2, j-1 : j+2]

        # --- ZADANIE ---
        # UZUPEŁNIJ WARUNEK (IF):
        # Sprawdź, czy w zmiennej "otoczenie" występuje minimum jedno 0 i minimum jedna 1

        if np.min(otoczenie) == 0 and np.max(otoczenie) == 1: # <--- W TO MIEJSCE STUDENT WPISUJE WARUNEK

            # Jeśli definicja matematyczna jest spełniona, oznaczamy ten punkt jako brzeg (1)
            brzeg_obiektu[i, j] = 1

print("Obliczanie brzegu zakończone!")
```

```python
# Rysowanie wyników
fig, axs = plt.subplots(1, 2, figsize=(12, 6))

# Wykres 1: Sam wyliczony brzeg
axs[0].imshow(brzeg_obiektu, cmap='Greys')
axs[0].set_title("Brzeg zbioru (Wyliczone kontury)")
axs[0].axis('off')

# Wykres 2: Brzeg nałożony na oryginalne zdjęcie
axs[1].imshow(zdjecie_male, cmap='gray', alpha=0.7)
# Tworzymy maskę, aby czerwony kolor wyświetlił się tylko tam, gdzie brzeg_obiektu == 1
maska = np.ma.masked_where(brzeg_obiektu == 0, brzeg_obiektu)
axs[1].imshow(maska, cmap='autumn', interpolation='none', alpha=1.0)
axs[1].set_title("Brzeg A nałożony na oryginał")
axs[1].axis('off')

plt.tight_layout()
plt.show()
```

# Zadanie 3: Zwartość w 3D – Dynamiczna "Sfera Bezpieczeństwa"
## Wprowadzenie teoretyczne

Na wykładzie zdefiniowaliśmy zwartość. W przestrzeni euklidesowej $\mathbb{R}^n$ (nawet dla $n=3$) zbiór jest zwarty, gdy jest domknięty i ograniczony.

Wizualizacja w 3D pozwala nam zobaczyć "chmurę pacjentów" jako obiekt przestrzenny. Nasz model topologiczny $\overline{B(\mathbf{x}_0,R)}$ będzie teraz pełną kulą w 3D. Pacjenci, którzy znajdą się poza tą bryłą, to przypadki nietypowe, których parametry (BMI, cholesterol, glukoza) wspólnie wykraczają poza normę.

##Cel zadania

Zaimplementujesz algorytm, który:

1. Przetworzy dane pacjentów w 3 wymiarach.

2. Automatycznie wyliczy promień R tak, aby 95% objętości danych znalazło się wewnątrz zwartej kuli.

3. Wygeneruje interaktywny wykres 3D z "bańką" bezpieczeństwa.

## Krok 1: Pobieranie danych medycznych

Wykorzystamy profesjonalny zbiór danych diabetes. Dane te są już wstępnie przetworzone (wyśrodkowane i przeskalowane), co jest standardem w uczeniu maszynowym. Pobieramy trzy cechy: BMI, S5 (trójglicerydy) oraz S6 (glukoza).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from mpl_toolkits.mplot3d import Axes3D

# --- 1. DANE 3D ---
diabetes = load_diabetes()
# Pobieramy 3 cechy: BMI, S5, S6 (indeksy 2, 8, 9)
X_3d = diabetes.data[:, [2, 8, 9]]
x0_3d = np.array([0, 0, 0]) # Środek w R^3

print(f"Załadowano dane 3D dla {X_3d.shape[0]} pacjentów.")
```

##Krok 2: Automatyczne definiowanie "Sfery Bezpieczeństwa" (Zbiór Zwarty)

Większość wyników medycznych pacjentów skupia się wokół środka układu współrzędnych (ponieważ nasze dane zostały wcześniej wyśrodkowane). Przyjmijmy, że środek naszej trójwymiarowej kuli to punkt x0​=[0,0,0].

Zamiast zgadywać sztywną wartość normy, musimy dobrać promień $R$ w sposób inteligentny. Wyznaczymy go tak, aby nasza domknięta kula (która zgodnie z twierdzeniem Heinego-Borela jest zbiorem zwartym) objęła dokładnie 95% najczęstszych przypadków. Pozostałe 5% pacjentów znajdzie się poza kulą i zostanie uznane za anomalie wymagające dokładniejszego zbadania przez lekarza.

Jak zaprogramować tę topologiczną selekcję?
Przełóż poniższe kroki matematyczne na kod:

1. Oblicz odległości: Dla każdego pacjenta z macierzy X_3d oblicz jego odległość euklidesową od środka x0_3d.
-Podpowiedź: Użyj funkcji `np.linalg.norm()`. Pamiętaj o dodaniu parametru `axis=1`, aby policzyć normę dla każdego wiersza osobno. Zapisz wynik w zmiennej wszystkie_odleglosci.

2. Posortuj wyniki: Użyj funkcji `np.sort()`, aby ułożyć wyliczone odległości rosnąco (od pacjentów najbliższych "ideałowi" do najbardziej nietypowych). Zapisz je jako odleglosci_posortowane.

3. Znajdź promień R: Oblicz indeks odpowiadający 95% danych (pomnóż całkowitą liczbę pacjentów przez 0.95 i rzutuj wynik na typ całkowity int, pamiętając o odjęciu 1 ze względu na indeksowanie od zera). Następnie wyciągnij z posortowanej tablicy odległość pod tym indeksem. To będzie Twój promień `R_wyliczone`.

4. Wyłap anomalie (Pętla for): Stwórz pustą listę anomalie_maska. Następnie napisz pętlę iterującą po wszystkich elementach tablicy wszystkie_odleglosci. Jeśli badana odległość jest ostro większa niż promień $R$, dodaj do listy wartość `True`, w przeciwnym razie dodaj `False`.


```python
# ==========================================
# GOTOWE ROZWIĄZANIE (KLUCZ DLA WYKŁADOWCY)
# ==========================================

# 1. Oblicz odległość euklidesową wszystkich pacjentów od środka układu
wszystkie_odleglosci = np.linalg.norm(X_3d - x0_3d, axis=1)

# 2. Posortuj odległości rosnąco
odleglosci_posortowane = np.sort(wszystkie_odleglosci)

# 3. Wyznacz indeks dla 95% populacji i pobierz wartość promienia
indeks_95 = int(0.95 * len(odleglosci_posortowane)) - 1
R_wyliczone = odleglosci_posortowane[indeks_95]
print(f"Wyliczony promień R dla 95% danych: {R_wyliczone:.4f}")

# 4. Stwórz maskę logiczną dla pacjentów za pomocą pętli FOR
anomalie_maska = []

# Iteracja po wszystkich wyliczonych dystansach
for dystans in wszystkie_odleglosci:
    if dystans > R_wyliczone:
        anomalie_maska.append(True)
    else:
        anomalie_maska.append(False)

# Konwersja listy na tablicę numpy (niezbędne, aby zadziałało na wykresie w kolejnym kroku)
anomalie_maska = np.array(anomalie_maska)

print(f"Liczba wykrytych anomalii (5%): {sum(anomalie_maska)}")
```

## Krok 3: Wizualizacja 3D

Użyjemy biblioteki `mpl_toolkits.mplot3d`, aby stworzyć przestrzenną sferę ograniczającą nasze dane. Nie umusisz tutaj nic edytować, uruchom jedynie komórkę.

```python
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# Rysujemy pacjentów
ax.scatter(X_3d[~anomalie_maska, 0], X_3d[~anomalie_maska, 1], X_3d[~anomalie_maska, 2],
           c='green', alpha=0.5, s=20, label='95% (Wewnątrz zbioru zwartego)')

ax.scatter(X_3d[anomalie_maska, 0], X_3d[anomalie_maska, 1], X_3d[anomalie_maska, 2],
           c='red', marker='x', s=50, label='5% Anomalii')

# Generujemy siatkę dla sfery (brzeg kuli w 3D)
u = np.linspace(0, 2 * np.pi, 30)
v = np.linspace(0, np.pi, 30)
x_sfera = R_wyliczone * np.outer(np.cos(u), np.sin(v))
y_sfera = R_wyliczone * np.outer(np.sin(u), np.sin(v))
z_sfera = R_wyliczone * np.outer(np.ones(np.size(u)), np.cos(v))

# Rysujemy półprzezroczystą sferę
ax.plot_surface(x_sfera, y_sfera, z_sfera, color='blue', alpha=0.1, linewidth=0)
ax.plot_wireframe(x_sfera, y_sfera, z_sfera, color='blue', alpha=0.1, label=r'Brzeg $\partial B(x_0, R)$')

# Stylizacja
ax.set_xlabel('BMI')
ax.set_ylabel('S5 (Trójglicerydy)')
ax.set_zlabel('S6 (Glukoza)')
ax.set_title(f'Topologia Medyczna 3D: Zbiór zwarty dla R={R_wyliczone:.4f}')
ax.legend()

plt.show()
```

# Krok 1: Algorytm sprawdzający przynależność do kuli

Twoim zadaniem jest napisać funkcję, która otrzymuje macierz punktów X i sprawdza, które z nich należą do kuli $B(0,1)$. Matematycznie punkt $x$ należy do kuli wpisanej, jeśli jego norma euklidesowa $\|x\| \leq 1$.

```python
import numpy as np
import matplotlib.pyplot as plt

def oblicz_procent_w_kuli(wymiar_n, liczba_punktow=100000):
    # Generujemy losowe punkty w n-wymiarowej kostce [-1, 1]^n
    # Wymiar macierzy: (liczba_punktow, wymiar_n)
    X = np.random.uniform(-1, 1, (liczba_punktow, wymiar_n))

    # --- ZADANIE ---
    # 1. Oblicz odległość (normę L2) każdego punktu od środka układu (0,0,...,0)
    # Podpowiedź: użyj np.linalg.norm z parametrem axis=1
    odleglosci = np.linalg.norm(X, axis=1)

    # 2. Sprawdź, które punkty są wewnątrz kuli (odległość <= 1)
    w_kuli = odleglosci <= 1

    # Obliczamy i zwracamy procent punktów w kuli
    procent = (np.sum(w_kuli) / liczba_punktow) * 100
    return procent

# Test dla 2D (spodziewamy się około 78.5%)
wynik_2d = oblicz_procent_w_kuli(wymiar_n=2)
print(f"Dla 2 wymiarów w kuli znalazło się {wynik_2d:.2f}% punktów.")
```

## Krok 2: Symulacja wzrostu wymiarów (Klątwa Wymiarowości)

W poprzednim kroku napisałeś funkcję `oblicz_procent_w_kuli`, która świetnie radzi sobie z wyliczaniem objętości kuli w dowolnym wymiarze. Teraz sprawdzimy, co się dzieje ze zbiorem ograniczonym, gdy zaczniemy wykładniczo zwiększać liczbę wymiarów przestrzeni (co w Machine Learningu odpowiada dodawaniu nowych cech do zbioru danych).

Przeanalizujemy wymiary od n=2 aż do n=15. Przygotuj się na dość szokujący wynik!

##Twoje zadanie:
Zaimplementuj pętlę, która przeprowadzi symulację i zbierze dane do naszego finałowego wykresu:

1. `. W kodzie poniżej przygotowaliśmy dla Ciebie tablicę `wymiary` oraz pustą listę `procenty`.

2. Stwórz pętlę for, która przejdzie przez każdy element n z tablicy wymiary.

3. Wewnątrz pętli wywołaj swoją funkcję `oblicz_procent_w_kuli(n)`.

4. Wynik zwrócony przez funkcję (czyli wyliczony procent) dodaj na koniec listy `procenty`.

5. Użyj funkcji `print()`, aby na bieżąco wyświetlać badany wymiar i jego wynik. Pomoże Ci to śledzić postępy symulacji na żywo.

```python
wymiary = np.arange(2, 16)
procenty = []

print("Przeprowadzanie symulacji... to może chwilę zająć.")
for n in wymiary:
    wynik = oblicz_procent_w_kuli(n)
    procenty.append(wynik)
    print(f"Wymiar {n:2d}: {wynik:6.2f}% objętości")

# --- WIZUALIZACJA ---
plt.figure(figsize=(10, 6))
plt.plot(wymiary, procenty, 'ro-', linewidth=2, markersize=8, label="Procent objętości w kuli")
plt.fill_between(wymiary, procenty, color='red', alpha=0.1)

plt.axhline(0, color='black', linewidth=1)
plt.title("Klątwa Wymiaru: Objętość kuli $B(0,1)$ względem kostki opisanej")
plt.xlabel("Liczba wymiarów ($n$)")
plt.ylabel("Procent punktów w kuli [%]")
plt.xticks(wymiary)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
